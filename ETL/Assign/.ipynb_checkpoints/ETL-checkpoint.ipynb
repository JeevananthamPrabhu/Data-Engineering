## Step 1:  Extract data from CSV

```python
import pandas as pd
df=pd.read_csv('Sales_data.csv')
print(df)
```

## Step2:  Transform
## Remove null values
## Calculate tax (Price * 1.18)

```python
df=df.dropna()
df['Price with tax']=df['Price']*1.18
print(df)
```

## Step3 : Load into SQL database

```python
import sqlite3 as sql
conn=sql.connect("sales_db")
df.to_sql("sales_table",conn,if_exists='replace',index=False)
```

```python
query="select * from sales_table"
sqldf=pd.read_sql(query,conn)
conn.close()
print(sqldf)
```
