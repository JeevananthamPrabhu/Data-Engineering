```python
import pandas as pd

# Load datasets
bank_df = pd.read_csv("bank.csv")
customer_df = pd.read_csv("customer_expense.csv")

# Rename long column names for easier use
bank_df = bank_df.rename(columns={
'Grocery_Cashback(%) [Zepto, BigBasket, etc.]':'grocery_cb',
'Travel_Cashback(%)':'travel_cb',
'Dining_Cashback [Swiggy, Zomato, etc.](%)':'dining_cb',
'Min_Income_Required(Yearly)':'min_income',
'Limit(montly)':'limit',
'Lounge access per year':'lounge',
'Rewards Points':'reward_points'
})

print("===================================")
print("1. Best Credit Card for Grocery")
print("===================================")

grocery_spend = 5000

bank_df["grocery_cashback"] = grocery_spend * bank_df["grocery_cb"]/100
print(bank_df.sort_values("grocery_cashback",ascending=False)[
["Bank_Name","Card_Name","grocery_cashback"]
].head())


print("===================================")
print("2. Cards with No Annual Fee")
print("===================================")

no_fee = bank_df[bank_df["Annual_Fee"] == 0]
print(no_fee[["Bank_Name","Card_Name","Annual_Fee"]])


print("===================================")
print("3. Cards Based on Customer Income")
print("===================================")

customer_income = 500000

eligible_cards = bank_df[bank_df["min_income"] <= customer_income]
print(eligible_cards[["Bank_Name","Card_Name","min_income"]])


print("===================================")
print("4. Total Cashback for Customer")
print("===================================")

grocery = 4000
travel = 2000
dining = 1500

bank_df["total_cashback"] = (
grocery * bank_df["grocery_cb"]/100 +
travel * bank_df["travel_cb"]/100 +
dining * bank_df["dining_cb"]/100
)

print(bank_df[["Bank_Name","Card_Name","total_cashback"]])


print("===================================")
print("5. Best Dining Cashback Cards")
print("===================================")

print(bank_df.sort_values("dining_cb",ascending=False)[
["Bank_Name","Card_Name","dining_cb"]
].head())


print("===================================")
print("6. Annual Fee vs Cashback")
print("===================================")

bank_df["net_value"] = bank_df["total_cashback"] - bank_df["Annual_Fee"]

print(bank_df.sort_values("net_value",ascending=False)[
["Bank_Name","Card_Name","net_value"]
].head())


print("===================================")
print("7. Filter Cards by Monthly Limit")
print("===================================")

required_limit = 50000

limit_cards = bank_df[bank_df["limit"] >= required_limit]
print(limit_cards[["Bank_Name","Card_Name","limit"]])


print("===================================")
print("8. Cards with Highest Reward Points")
print("===================================")

print(bank_df.sort_values("reward_points",ascending=False)[
["Bank_Name","Card_Name","reward_points"]
].head())


print("===================================")
print("9. Cards with Lounge Access")
print("===================================")

lounge_cards = bank_df[bank_df["lounge"] > 0]

print(lounge_cards[["Bank_Name","Card_Name","lounge"]])


print("===================================")
print("10. Credit Card Recommendation")
print("===================================")

income = 500000
grocery = 4000
travel = 2000
dining = 1500

eligible = bank_df[bank_df["min_income"] <= income]

eligible["cashback"] = (
grocery * eligible["grocery_cb"]/100 +
travel * eligible["travel_cb"]/100 +
dining * eligible["dining_cb"]/100
)

eligible["net_benefit"] = eligible["cashback"] - eligible["Annual_Fee"]

recommended = eligible.sort_values("net_benefit",ascending=False)

print(recommended[
["Bank_Name","Card_Name","net_benefit"]
].head(3))
```
