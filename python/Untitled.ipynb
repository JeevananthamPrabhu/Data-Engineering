```python
import pandas as pd
```

```python
bank_df=pd.read_csv("bank.csv")
customer_df=pd.read_csv("customer_expense.csv")
```

```python
max_gro=bank_df["Grocery_Cashback"].max()
best_grocery=bank_df[bank_df["Grocery_Cashback"]==max_gro][["Bank_Name","Card_Name","Grocery_Cashback","CardType"]]
best_grocery
```

```python
no_annual = bank_df[(bank_df["Annual_Fee"] == 0) & (bank_df["CardType"] == "Credit")][["Bank_Name","Card_Name","Annual_Fee"]]
no_annual
```

```python
bank_df["Min_Income_Required"]= bank_df["Min_Income_Required"].str.replace(",","").astype(int)
```

```python
merged=pd.merge(customer_df,bank_df[bank_df['CardType']=='Credit'],how='cross')
elg=merged[merged['total Income']>=merged['Min_Income_Required']]
elg[['Name','Bank_Name','Card_Name','Min_Income_Required','total Income']]
```

```python
merged=pd.merge(customer_df,bank_df[bank_df['CardType']=='Credit'],how='cross')
elg=merged[merged['total Income']>=merged['Min_Income_Required']]
elg['Total_Cashback']=(elg['Travelling']*elg['Travel_Cashback']/100+
elg['Grocery']*elg['Grocery_Cashback']/100+
elg['Dining']*elg['Dining_Cashback']/100)
cashback=elg[elg['Total_Cashback']!=0][['Name','Bank_Name','Card_Name','Total_Cashback']]
cashback
```

```python
max_din=bank_df["Dining_Cashback"].max()
best_din=bank_df[bank_df["Dining_Cashback"]==max_gro][["Bank_Name","Card_Name","Dining_Cashback"]]
best_din
```

```python
merged=pd.merge(customer_df,bank_df[bank_df['CardType']=='Credit'],how='cross')
elg=merged[merged['total Income']>=merged['Min_Income_Required']]
elg['Total_Cashback']=(elg['Travelling']*elg['Travel_Cashback']/100+
elg['Grocery']*elg['Grocery_Cashback']/100+
elg['Dining']*elg['Dining_Cashback']/100)
elg['benifit']=elg['Total_Cashback']-elg['Annual_Fee']
benifit=elg[['Name','Bank_Name','Card_Name','benifit']]
benifit
```

```python
bank_df["Limit"]= bank_df["Limit"].str.replace(',','').astype(int)
```

```python
customer_df['Monthly_Spend']=(customer_df['Grocery']+customer_df['Travelling']+customer_df['Dining']+customer_df['Electricity']
+customer_df['Internet']+customer_df['Entertainment']+customer_df['EMI']+customer_df['OTT']
+customer_df['Ecommerce']+customer_df['UPI_Transactions'])
merged=pd.merge(customer_df,bank_df[bank_df['CardType']=='Credit'],how='cross')
elg=merged[merged['total Income']>=merged['Min_Income_Required']]
suitable=elg[elg['Limit']>=elg['Monthly_Spend']][['Name','Bank_Name','Card_Name','Limit','monthly_spend']]
suitable
```
